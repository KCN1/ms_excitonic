Second, while short-range exchange and overlap interactions decay exponentially and are tightly restricted to the nearest neighbors, long-range Coulombic coupling scales as $1/r$ and theoretically necessitates the inclusion of next-nearest-neighbour (NNN) terms. Finally, the one-dimensional nearest-neighbour approximation intrinsically neglects inter-stack interactions. In a realistic molecular aggregate, each ZnPc unit is surrounded by multiple adjacent neighbors across neighboring columns.
