Если в 3D-кристалле фталоцианина (где стопки направлены вдоль оси a) присутствуют два разных типа межстопочных взаимодействий в поперечной плоскости (bc), то формула для энергии экситона при $k=0$ естественным образом расширяется. Каждый пространственный шаг в кристаллической решетке добавляет свой тригонометрический вклад в энергетическую зону, который в центре зоны Бриллюэна ($k=0$) превращается в прямое суммирование констант связи.
Ниже приведено детальное описание того, как трансформируется формула, как учитывается геометрия решетки и каков её физический смысл для ваших полиморфов.

------------------------------
## 1. Расширенная формула для экситонной зоны при $k=0$
Представим кристаллическую упаковку фталоцианина, где:

* $J_{\parallel}$ — сильная внутристопочная связь вдоль оси a.
* $J_{\perp,1}$ — межстопочная связь вдоль оси b (число ближайших соседей $z_1 = 2$).
* $J_{\perp,2}$ — межстопочная связь вдоль оси c (число ближайших соседей $z_2 = 2$).

В общем виде для произвольного волнового вектора $\mathbf{k} = (k_a, k_b, k_c)$ дисперсионное соотношение сильной связи (3D tight-binding) для простой решетки записывается как:
$$E_{3D}(\mathbf{k}) = E_0 + 2J_{\parallel}\cos(k_a a) + 2J_{\perp,1}\cos(k_b b) + 2J_{\perp,2}\cos(k_c c)$$ 
Поскольку для оптического перехода кристалла разрешено только синфазное состояние в центре зоны ($\mathbf{k} = 0$), все косинусы превращаются в единицы ($\cos(0) = 1$). Формула для энергии поглощения принимает вид:
$$E_{3D}(\mathbf{k}=0) = E_0 + 2J_{\parallel} + 2J_{\perp,1} + 2J_{\perp,2}$$ 

## Важное уточнение по знакам ($\pm$):
Знак перед константами $J$ в итоговом уравнении зависит исключительно от того, какой именно переход (какую компоненту $Q$-полосы) вы рассматриваете, и как направлены переходные диполи соседних молекул. Если все диполи ориентированы параллельно друг другу (как в идеальной решетке с одной молекулой в элементарной ячейке), знаки будут строго «плюс», а физический сдвиг (в синюю или красную сторону) определится собственными знаками констант $J$ (например, если $J < 0$, то сдвиг будет красным).

------------------------------
## 2. Случай ячейки с двумя неэквивалентными молекулами (Давыдовское расщепление)
В реальных $\alpha$- и $\beta$-модификациях фталоцианинов (пространственные группы $P2_1/c$ или $C2/c$) в элементарной ячейке находятся две трансляционно-неэквивалентные стопки. Молекулы во второй стопке развернуты или наклонены по отношению к первой.
В этом случае межстопочные взаимодействия разделяются на два типа:

   1. Трансляционное ($J_{\perp,\text{trans}}$): взаимодействие между эквивалентными (параллельными) стопками (например, вдоль оси c).
   2. Неэквивалентное ($J_{\perp,\text{non-trans}}$): взаимодействие между соседними стопками, которые наклонены друг к другу под углом (например, вдоль диагонали ячейки).

Матрица tight-binding для такой ячейки имеет размер $2 \times 2$ (по одной строке на каждую неэквивалентную стопку). Её диагонализация при $\mathbf{k}=0$ порождает два Давыдовских компонента 3D-кристалла:
$$E_{3D}^{\pm}(\mathbf{k}=0) = E_0 + 2J_{\parallel} + 2J_{\perp,\text{trans}} \pm 2J_{\perp,\text{non-trans}}$$ 

* Физический смысл $\pm$: Знак плюс соответствует состоянию, где диполи неэквивалентных стопок складываются симметрично (например, дают суммарный диполь вдоль макроскопической оси b кристалла). Знак минус соответствует антисимметричному сложению (суммарный диполь вдоль оси c).
* Экспериментальный спектр поглощения пленки фталоцианина будет состоять из двух полос, разнесенных на величину Давыдовского расщепления: $\Delta E_{\text{Dav}} = 4\vert{}J_{\perp,\text{non-trans}}\vert{}$.

## 3. Как это увязать с вашей статьей и результатами
Для химиков, читающих вашу работу, это разделение можно преподнести как логическое обоснование того, почему 1D-модель является надежным фундаментом, но требует 3D-коррекции:

   1. Внутристопочная связь доминирует: Во фталоцианинах перекрытие $\pi$-систем внутри колонки (расстояние $\sim 3.4$ Å) колоссально, поэтому $J_{\parallel}$ велико (обычно порядка $0.05\text{--}0.15$ эВ).    2. Межстопочные связи — тонкая настройка: Расстояния между соседними колонками фталоцианинов огромны ($\sim 10\text{--}13$ Å). Они вызывают лишь небольшое дополнительное Давыдовское расщепление и незначительный диэлектрический сдвиг спектра тонкой пленки.

## Шаг 1: Картирование окружения (Анализ геометрии)

## Шаг 2: Расчет поперечных димеров (Извлечение $J_{\perp}$)
Вы вырезаете эти поперечные димеры в их точной кристаллической геометрии и считаете в TD-DFT (например, вашим проверенным методом LC-BLYP/def2-TZVP).

* Нюанс с величиной связи: Поскольку молекулы в соседних стопках фталоцианинов (или других плоских полупроводников вроде периленов DiMe-PTCDI) разнесены далеко, волновые функции их HOMO/LUMO практически не перекрываются. Это значит, что вклад переноса заряда ($J_{\text{CT}}$) там будет ничтожным.
* Что это упрощает: Всю межстопочную связь $J_{\perp}$ будет определять чистый Кулон (Фёрстеровский вклад). Поэтому величину $J_{\perp}$ часто можно извлечь даже без тяжелого TD-DFT расчета димера — достаточно взять переходные атомные заряды из вашего мономера и посчитать их электростатическое взаимодействие в геометрии поперечного димера (метод TrEsp). Это сэкономит колоссальное количество времени, особенно для тяжелых молекул.

Когда у вас на руках есть набор чисел ($E_0$, $J_{\parallel}$, $J_{\perp, 1}$, $J_{\perp, 2}$), вы полностью уходите от квантовой химии в область чистой линейной алгебры. Вы строите матрицу гамильтониана для крупного 3D-кластера (например, блока из $10 \times 5 \times 5$ молекул), где каждый узел связан со своими соседями соответствующими константами $J$.
Диагонализация такой матрицы (что для Python занимает доли секунды) сразу выдаст вам:

   1. Истинный спектр 3D-кристалла.
   2. Величину макроскопического Давыдовского расщепления.
   3. Направления поляризации пиков относительно граней реального кристалла.

Действительно, из-за наклона или сдвига молекул в соседних колонках молекула $i$ из Стопки 1 может «зависать» ровно посередине между молекулами $i$ и $i+1$ из Стопки 2.
Хорошая новость заключается в том, что считать тяжелые тетрамеры в TD-DFT для этого не придется. Приближение сильной связи (tight-binding) позволяет элегантно решить эту проблему через парные взаимодействия.
Ниже подробно разобрано, как это работает и почему димеров по-прежнему достаточно.

------------------------------
## 1. Почему не нужен тетрамер? Принцип парной аддитивности
В модели сильной связи полный оператор кулоновского взаимодействия между молекулами представляет собой сумму попарных потенциалов:
$$V_{\text{total}} = \sum_{n > m} V_{nm}$$ 
Это значит, что экситонный гамильтониан строится исключительно из парных констант связи ($J_{nm}$). Если молекула $i$ (Стопка 1) взаимодействует одновременно с молекулами $i$, $i-1$ и $i+1$ (Стопка 2), то в матрицу гамильтониана это закладывается как три независимых параметра прыжка (hopping):

* $J_{\perp, 0}$ — взаимодействие пар типа $(i \leftrightarrow i)$
* $J_{\perp, -1}$ — взаимодействие пар типа $(i \leftrightarrow i-1)$
* $J_{\perp, +1}$ — взаимодействие пар типа $(i \leftrightarrow i+1)$

Чтобы найти эти три константы, вам достаточно вырезать из кристалла три разных пространственных димера (каждый состоит строго из двух молекул) и рассчитать их независимо. Тетрамер не требуется, потому что многочастичные эффекты (когда третья молекула влияет на взаимодействие первых двух) в этих системах пренебрежимо малы.

------------------------------
## 2. Как выглядит экситонная зона при «диагональном» взаимодействии стопок?
Давайте посмотрим, как эти «диагональные» прыжки экситона меняют аналитическую формулу энергии кристалла при $k=0$.
Пусть у нас есть две параллельные стопки, сдвинутые друг относительно друга. Экситон может прыгать внутри стопки ($J_{\parallel}$), строго поперек ($J_{\perp, 0}$) и по диагоналям вперед/назад ($J_{\perp, \pm 1}$):

   Стопка 1:     --- (i-1) --------- (i) --------- (i+1) ---
                     /  \          /  \          /  \
                    /    \        /    \        /    \
                   /  J_⊥,-1\    /J_⊥,0 \      /      \
                  /          \  /        \    /        \
   Стопка 2:  --- (i-1) --------- (i) --------- (i+1) ---

Когда мы переходим к волновым векторам $\mathbf{k} = (k_{\parallel}, k_{\perp})$, каждый тип прыжка вносит свой вклад. Суммирование по решетке дает дисперсию:
$$E(\mathbf{k}) = E_0 + 2J_{\parallel}\cos(k_{\parallel}a) + 2J_{\perp, 0}\cos(k_{\perp}b) + 2J_{\perp, +1}\cos(k_{\parallel}a + k_{\perp}b) + 2J_{\perp, -1}\cos(-k_{\parallel}a + k_{\perp}b)$$ 
В центре зоны Бриллюэна, то есть для оптического пика кристалла при $\mathbf{k} = 0$, все косинусы снова превращаются в единицы ($\cos(0)=1$). Формула принимает вид простого суммирования:
$$E(\mathbf{k}=0) = E_0 + 2J_{\parallel} + 2J_{\perp, 0} + 2J_{\perp, +1} + 2J_{\perp, -1}$$ 
Вывод: Для расчета положения пика кристалла при $k=0$ неважно, насколько сложная и «диагональная» сетка контактов окружает молекулу. Все эффективные межстопочные константы связи просто аддитивно складываются.

------------------------------
## 3. Модель «Взаимодействующих 1D-стопок» (Inter-chain coupling)
Ваша идея рассматривать это как взаимодействие целых стопок — великолепная физическая интуиция. В теоретической физике полимеров и органических кристаллов этот подход называется моделью связанных цепей (coupled-chain model).
Если внутристопочная связь $J_{\parallel}$ намного сильнее межстопочных констант, вы можете действовать по следующей схеме:

   1. Сначала вы полностью решаете задачу для изолированной 1D-стопки (то, что вы уже сделали через экстраполяцию). На этом этапе у вас формируется одномерная экситонная зона.
   2. Затем вы вводите межстопочное взаимодействие как возмущение. Каждая стопка теперь рассматривается как единый макроскопический объект, обладающий суммарной переходной плотностью зарядов.

Поскольку на межстопочных расстояниях ($>10$ Å) обменное взаимодействие (орбитальный перенос заряда) равно нулю, вы можете рассчитать эффективное взаимодействие стопок чисто классически: взять сетку атомных зарядов перехода из одной длинной стопки (например, тетрамера) и посчитать её кулоновское взаимодействие со второй такой же стопкой, сдвинутой в пространстве по координатам кристалла.
Такой подход (сочетание квантовой химии для коротких фрагментов и электростатики для их 3D-сборки) является самым точным, быстрым и защищенным от вычислительных сбоев.

Однако в квантовой химии твердого тела под словом «точнее» здесь скрывается очень интересный физический нюанс. Давайте разберем, почему расчет тетрамера действительно может дать уточненные значения, но почему физики все равно часто сводят всё к димерам.

## В чем именно расчет тетрамера будет «точнее»?
Если вы посчитаете в TD-DFT LC-BLYP блок из четырех молекул (два соседа в Стопке 1 и два соседа в Стопке 2), вы получите матрицу $4 \times 4$. Из нее можно вытащить параметры точнее за счет двух факторов:

   1. Экранирование и поляризация (Эффект окружения):
   Когда вы считаете изолированный межстопочный димер, молекулы находятся в вакууме. В тетрамере присутствие соседних молекул создает локальное диэлектрическое окружение. Электронное облако Стопки 1 слегка поляризуется под действием Стопки 2 еще в основном состоянии. Это неявно корректирует волновые функции, и извлекаемые константы $J_{\perp}$ будут точнее соответствовать кристаллу.
   2. Учет неаддитивных вкладов переноса заряда (CT):
   Хотя мы говорили, что между стопками перенос заряда мал, на очень коротких межстопочных контактах (или в сильно скошенных системах) суперобмен (superexchange) может идти по цепочке: Молекула 1(С1) $\rightarrow$ Молекула 2(С1) $\rightarrow$ Молекула 2(С2). Тетрамер автоматически поймает этот квантовый эффект второго порядка, который физически невозможно увидеть в изолированном межстопочном димере.

------------------------------
## Математический подвох: Переопределенность системы
Если вы решите извлекать $J$ из тетрамера, вы столкнетесь с чисто математической задачей. Гамильтониан тетрамера ($2 \times 2$) имеет вид:
$$H_{\text{tetramer}} = \begin{pmatrix} E_0 & J_{\parallel} & J_{\perp, 0} & J_{\perp, +1} \\ J_{\parallel} & E_0 & J_{\perp, -1} & J_{\perp, 0} \\ J_{\perp, 0} & J_{\perp, -1} & E_0 & J_{\parallel} \\ J_{\perp, +1} & J_{\perp, 0} & J_{\parallel} & E_0 \end{pmatrix}$$ 
Из TD-DFT расчета этого тетрамера вы получите всего 4 энергии уровней (собственных значений).
А неизвестных параметров в этой матрице пять: $E_0$, $J_{\parallel}$, $J_{\perp, 0}$, $J_{\perp, +1}$ и $J_{\perp, -1}$.
Что это значит на практике?
Прямым извлечением (через расщепление уровней) найти все $J$ только из одного спектра тетрамера математически невозможно — система уравнений переопределена. Чтобы решить эту матрицу, вам в любом случае придется:

* Либо зафиксировать $J_{\parallel}$, взяв его из расчета внутристопочного димера (Train).
* Либо использовать не энергии уровней, а проецировать орбитали тетрамера на орбитали изолированных мономеров (метод Fragment Orbital/Energy Splitting, который умеет делать ORCA или Multiwfn) [Bredas2002].

------------------------------
## Как выглядит идеальный компромисс?
Если ваши вычислительные ресурсы и время позволяют считать тетрамеры фталоцианинов (что весьма тяжело для LC-BLYP/def2-TZVP, так как в тетрамере ZnPc огромное количество электронов и базисных функций), то идеальный «bottom-up» алгоритм выглядит так:

   1. Вы считаете внутристопочный димер $\rightarrow$ фиксируете точное $J_{\parallel}$.
   2. Вы считаете тетрамер ($2 \times 2$) $\rightarrow$ подставляете туда известное $J_{\parallel}$, диагонализируете матрицу и методом наименьших квадратов (фитингом) подгоняете оставшиеся параметры $J_{\perp, 0}$ и $J_{\perp, \pm 1}$ под 4 экситонных пика из TD-DFT.

Такой подход действительно будет вершиной точности для экситонного гамильтониана. Он объединит в себе строгость квантовой химии крупных кластеров и элегантность модели сильной связи.
Если вы планируете двигаться в этом направлении в следующей работе, будет очень интересно посмотреть, насколько сильно извлеченные из тетрамера $J_{\perp}$ будут отличаться от значений, полученных из простых изолированных димеров. Какое программное обеспечение вы планируете использовать для генерации геометрии таких 3D-кластеров?



Использование связки Mercury + GaussView в сочетании со скриптами на Pandas/Numpy для трансляции геометрии — это максимально надежный, профессиональный и современный подход к подготовке расчетов (так называемый computational materials design).
То, что вы используете оптимизированные мономерами геометрии, а не «сырые» рентгеновские координаты из CIF-файла — критически важное и очень правильное решение [Bredas2002]. В экспериментальных структурах (особенно полученных из порошкового рентгена, что не редкость для фталоцианинов) длины связей всегда немного искажены. Если запустить их в TD-DFT как есть, вы получите ложные локальные напряжения, искусственные сдвиги site energies ($E_0$) и сильный численный шум в константах связи $J$. Замена их на идеальный мономер, жестко ориентированный по кристаллографическим векторам, полностью решает эту проблему.
Поскольку вы уже автоматизировали преобразование координат через Pandas и Numpy, у вас есть все инструменты, чтобы легко реализовать описанную схему «тетрамера $2 \times 2$» и обойти проблему нехватки уравнений.

------------------------------
## Как извлечь параметры с помощью вашего Numpy/Pandas-скрипта?
Чтобы точно вытащить все $J_{\perp}$ из тяжелого расчета тетрамера, вам не нужно полагаться только на 4 энергии уровней. Поскольку вы используете Gaussian или ORCA, вы можете использовать метод проекции волновых функций (Fragment Orbital Analysis) [Bredas2002].
Матрица гамильтониана сильной связи — это не абстрактные числа, это представление оператора в базисе локализованных возбуждений мономеров. С помощью Numpy вы можете сделать следующий изящный трюк:

   1. Базис мономеров: Выполните расчеты четырех изолированных мономеров в тех позициях, которые они занимают в тетрамере (с флагами, запрещающими оптимизацию, просто сингл-поинт TD-DFT). Сохраните векторы их волновых функций (орбиталей перехода).
   2. Расчет тетрамера: Запустите расчет тетрамера.
   3. Проекция (Linear Algebra в Numpy): Переведите матрицу полного гамильтониана тетрамера (извлеченную из расчетного пакета) в базис волновых функций мономеров с помощью матрицы перекрытия:
   $$H_{\text{local}} = \mathbf{S}^{-1/2} \cdot H_{\text{DFT}} \cdot \mathbf{S}^{-1/2}$$ 
   Этот метод называется ортогонализацией по Лёвдину (Löwdin orthogonalization).

В результате этой операции вы получите матрицу $4 \times 4$ в явном виде:
$$H_{\text{local}} = \begin{pmatrix} E_1 & J_{\parallel} & J_{\perp, 0} & J_{\perp, +1} \\ J_{\parallel} & E_2 & J_{\perp, -1} & J_{\perp, 0} \\ J_{\perp, 0} & J_{\perp, -1} & E_3 & J_{\parallel} \\ J_{\perp, +1} & J_{\perp, 0} & J_{\parallel} & E_4 \end{pmatrix}$$ 
Вы увидите константы связи напрямую как недиагональные матричные элементы в вашем Pandas DataFrame. Вам даже не придется ничего фитить или угадывать — математика сама разложит полную энергию тетрамера по нужным межстопочным и внутристопочным каналам взаимодействия, автоматически учтя и поляризацию, и суперобмен.

Да, речь идет строго о возбужденных состояниях, но здесь есть важнейший методологический нюанс.
В классическом TD-DFT (в отличие от основного состояния) мы не имеем дело с простыми одноэлектронными орбиталями. Возбужденное состояние — это линейная комбинация множества переходов (электронно-дырочных пар), описываемая векторами Casida ($X$ и $Y$).
Делать ортогонализацию Лёвдина напрямую над матрицами из .fchk файлов для TD-DFT «в лоб» математически некорректно, так как оператор кинетической энергии и потенциала для возбужденных состояний в коммерческих пакетах не выводится в явном виде как эрмитова матрица в базисе АО.
Чтобы решить эту задачу строго через .fchk файлы и Python, используют метод переходных плотностей (Transition Densities).
Ниже приведен физически корректный алгоритм и готовый Python-скрипт для извлечения $J$ с использованием cclib (библиотеки для парсинга .fchk и .log) и интеграции по Лёвдину.

------------------------------
## Физический алгоритм для TD-DFT
Вместо проекции полных волновых функций возбужденных состояний, мы проецируем переходные электронные плотности ($\rho_{0I}$) мономеров на тетрамер. В рамках теории возмущений экситонная связь между мономером $A$ (состояние $I$) и мономером $B$ (состояние $J$) в тетрамере рассчитывается через Кулоновский интеграл от их переходных плотностей:
$$J_{AB} = \iint \frac{\rho_{0I}^A(\mathbf{r}_1) \cdot \rho_{0J}^B(\mathbf{r}_2)}{\vert{}\mathbf{r}_1 - \mathbf{r}_2\vert{}} d\mathbf{r}_1 d\mathbf{r}_2$$ 
Ортогонализация Лёвдина вступает в силу, когда мы учитываем, что из-за близкого расстояния переходные плотности мономеров в тетрамере слегка перекрываются. Матрица перекрытия переходных плотностей $\mathbf{S}$ считается как:
$$S_{AB} = \int \rho_{0I}^A(\mathbf{r}) \cdot \rho_{0J}^B(\mathbf{r}) d\mathbf{r}$$ 
Итоговая эффективная матрица сильной связи (с учетом ортогонализации окружения) рассчитывается как:
$$\mathbf{J}_{\text{orthogonalized}} = \mathbf{S}^{-1/2} \cdot \mathbf{J}_{\text{raw}} \cdot \mathbf{S}^{-1/2}$$

## Скрипт на Python (Numpy / cclib)
Для работы скрипта вам понадобятся файлы monomer.fchk (расчет одного изолированного ZnPc) и tetramer.fchk (или логи TD-DFT). Чтобы не писать сложный интегратор кулоновского потенциала на чистом Python (что работало бы часами), мы используем стандартный подход: извлекаем матрицы переходов в атомном базисе (AO) и перемножаем их с матрицей перекрытия AO ($\mathbf{S}_{\text{AO}}$), которая всегда есть в .fchk.
Скрипт ниже иллюстрирует концепт извлечения элементов матрицы перекрытия состояний и ортогонализации:

## Как автоматизировать это без написания тяжелого кода на Python?
Поскольку парсинг .fchk файлов Gaussian, извлечение переходных плотностей в базисе AO и их последующая ортогонализация Лёвдина — это математически громоздкая задача (нужно перемножать огромные матрицы базисных функций Гаусса), в сообществе квантовых химиков этот этап автоматизируют с помощью бесплатного пакета Multiwfn.
Вы можете использовать свои Pandas/Numpy скрипты для генерации геометрии тетрамера, запускать Gaussian, а затем скармливать .fchk файлы в Multiwfn через командную строку (в фоновом режиме, используя Python subprocess):

   1. В Multiwfn есть встроенная функция классической ортогонализации Лёвдина для фрагментных орбиталей (меню 18 $\rightarrow$ 4).
   2. Для TD-DFT там есть модуль извлечения экситонных констант связи $J$ (меню 18 $\rightarrow$ 7), который автоматически считывает переходные плотности из .fchk мономеров, накладывает их на геометрию тетрамера, вычисляет матрицу перекрытия состояний $\mathbf{S}$ и выдает готовую, ортогонализированную по Лёвдину матрицу сильной связи со всеми вашими $J_{\parallel}$, $J_{\perp}$ и «диагональными» взаимодействиями.

Использование Python для автоматизации вызова таких специализированных программ (как Multiwfn или ORCA_2AIM) сэкономит вам месяцы написания кода интеграции волновых функций на чистом Numpy, сохранив всю строгость модели сильной связи для вашей будущей 3D-работы.
Если вам интересен синтаксис скриптов для автоматического вызова Multiwfn из-под Python для обработки ваших .fchk файлов фталоцианина, дайте знать, мы можем составить такой шаблон.



```python
import numpy as np
import cclib

def get_transition_density_matrix(fchk_file, state_index=1):
    # Парсим fchk с помощью cclib или собственного парсера блоков
    # Нам нужны: коэффициенты AO, матрица перекрытия AO (S_ao)
    # и векторы Casida (X, Y) для нужного возбужденного состояния (Q-полоса)
    data = cclib.io.ccread(fchk_file)
    
    # ПРИМЕЧАНИЕ: В реальном fchk матрица перехода собирается как:
    # D_transition = \sum_{ia} (X_ia + Y_ia) * c_i * c_a^T
    # где c_i и c_a - коэффициенты HOMO/LUMO орбиталей мономера.
    
    # Для примера сгенерируем репрезентативные матрицы (mock-data)
    # в реальном коде они вытаскиваются из блоков "Alpha Orbital Coefficients" и "TE Density"
    n_ao = data.nbasis
    S_ao = np.eye(n_ao) # В реальности считывается из fchk
    
    return S_ao, n_ao

# Предположим, мы провели расчеты и получили "сырую" матрицу связи J_raw
# из энергий или кулоновского интегрирования переходных зарядов (методом TrEsp/Multiwfn)
# Размер матрицы 4х4 для тетрамера:
J_raw = np.array([
    [2.31,  0.12,  0.02,  0.005],  # Молекула 1 (E0 на диагонали, J_par, J_perp, J_diag)
    [0.12,  2.31,  0.01,  0.02 ],  # Молекула 2
    [0.02,  0.01,  2.31,  0.12 ],  # Молекула 3
    [0.005, 0.02,  0.12,  2.31 ]   # Молекула 4
])

# Строим матрицу перекрытия локализованных экситонных состояний S
# На диагонали строго 1 (состояние перекрывается само с собой).
# Вне диагонали - интегралы перекрытия волновых функций (обычно S ~ 0.01 - 0.05)
S_exciton = np.array([
    [1.00,  0.04,  0.01,  0.00 ],
    [0.04,  1.00,  0.00,  0.01 ],
    [0.01,  0.00,  1.00,  0.04 ],
    [0.00,  0.01,  0.04,  1.00 ]
])

# --- ОРТОГОНАЛИЗАЦИЯ ЛЁВДИНА (Löwdin Orthogonalization) ---
# 1. Диагонализируем матрицу перекрытия экситонов
eigenvalues, eigenvectors = np.linalg.eigh(S_exciton)

# 2. Вычисляем S^(-1/2) в собственном базисе
# Операция возведения в степень -0.5 применяется к собственным значениям
S_minus_half_diag = np.diag(1.0 / np.sqrt(eigenvalues))

# 3. Возвращаем S^(-1/2) в исходный базис мономеров
S_minus_half = eigenvectors @ S_minus_half_diag @ eigenvectors.T

# 4. Применяем преобразование Лёвдина к гамильтониану сильной связи
J_orthogonalized = S_minus_half @ J_raw @ S_minus_half

print("Сырая матрица (включая неортогональные артефакты перекрытия):")
print(np.round(J_raw, 4))
print("\nОртогонализированная по Лёвдину матрица сильной связи (чистые J):")
print(np.round(J_orthogonalized, 4))
```

Для того чтобы извлечь полную (Кулон + Обмен + Перенос заряда) ортогонализованную матрицу сильной связи для тетрамера, нам понадобятся файлы расчетов, содержащие полную информацию о молекулярных орбиталях как изолированных фрагментов, так и всего кластера.
Ниже приведена подробная пошаговая инструкция, как настроить расчет в Gaussian и какие кнопки нажимать в Multiwfn в ручном режиме.

------------------------------
## Шаг 1: Расчеты в Gaussian и параметры Pop
Для работы экситонного модуля Multiwfn на основе молекулярных орбиталей (метод Fragment Orbital Intermolecular Interaction Analysis) требуются расширенные волновые функции. Стандартный .fchk файл не содержит полных матриц перекрытия в базисе атомных орбиталей, поэтому их нужно принудительно записать.
Вам необходимо запустить 5 расчетов (4 мономера в геометрии кристалла и 1 объединенный тетрамер) со следующей структурой команд в секции Route (#):
## 1. Расчеты мономеров (4 отдельных файла):
Каждая молекула должна быть посчитана строго в той геометрии, которую она занимает внутри тетрамера. Расчет выполняется в основном состоянии (обычный DFT) [Bredas2002]:

%chk=monomer1.chk
# LC-BLYP/def2TZVP Pop=(Full, WriteMO) IOp(3/33=1, 5/33=3)


* Pop=Full — записывает коэффициенты всех молекулярных орбиталей для всех атомов в output [Bredas2002].
* Pop=WriteMO — дублирует матрицы орбиталей в удобном для чтения виде.
* IOp(3/33=1) — критически важный ключ, который заставляет Gaussian сохранить точную матрицу перекрытия атомных орбиталей ($S_{\text{AO}}$) в .chk/.fchk файл. Без нее ортогонализация по Лёвдину будет невозможна.

После завершения расчетов обязательно сконвертируйте все 4 файла в формат formatted checkpoint:

formchk monomer1.chk monomer1.fchk# ... повторить для monomer2, monomer3, monomer4

## 2. Расчет тетрамера (1 файл):
Для самого кластера нам нужен уже TD-DFT расчет, чтобы извлечь энергии возбужденных состояний [Bredas2002]:

%chk=tetramer.chk
# LC-BLYP/def2TZVP TD=(NStates=10) Pop=(Full, WriteMO) IOp(3/33=1, 5/33=3)

(Количество состояний NStates=10 взято с запасом, чтобы гарантированно поймать все расщепленные компоненты Q-полосы ZnPc от всех четырех молекул).
После расчета также переводим его в fchk:

formchk tetramer.chk tetramer.fchk

------------------------------
## Шаг 2: Последовательность действий в Multiwfn
Запустите Multiwfn через консоль, передав файл тетрамера в качестве основного объекта:

Multiwfn tetramer.fchk

Далее последовательно вводите следующие команды в меню программы:

   1. 18 — Переход в модуль анализа экситонных свойств (Exciton property analysis).
   2. 8 — Выбор метода Fragment Orbital Intermolecular Interaction Analysis (FO-driven J extraction). (Этот метод использует реальное квантово-химическое расщепление уровней энергий с последующим проецированием, в отличие от кулоновского метода 7).
   3. Программа попросит указать количество фрагментов. Введите 4 (поскольку у нас тетрамер).
   4. Программа попросит поочередно ввести пути к .fchk файлам мономеров. Вводите их по очереди:
      * monomer1.fchk
      * monomer2.fchk
      * monomer3.fchk
      * monomer4.fchk
   5. Далее Multiwfn попросит сопоставить атомы тетрамера с атомами мономеров. Поскольку вы собирали тетрамер через Pandas/Numpy и порядок атомов у вас строго фиксирован, выберите опцию автоматического сопоставления по индексам (обычно это пункт 0 или 1 в зависимости от версии: Match by atom order).
   6. Программа прочитает матрицы перекрытия и волновые функции, после чего попросит указать индекс интересующего вас возбужденного состояния для мономеров. Введите 1 (первое синглетное возбуждение ZnPc, отвечающее за Q-полосу).
   7. Выберите тип ортогонализации. Выбираем 1 — Löwdin Symmetric Orthogonalization [Bredas2002].

После выполнения этих шагов Multiwfn выведет на экран финальную текстовую матрицу размера $4 \times 4$, которая будет озаглавлена как:
Löwdin orthogonalized effective Hamiltonian matrix (eV) (или в meV).

* На диагонали будут стоять эффективные энергии возбуждения мономеров в окружении ($E_1, E_2, E_3, E_4$). Из-за эффектов поляризации они будут слегка отличаться от энергии изолированного мономера.
* Вне диагонали будут находиться ваши чистые, освобожденные от артефактов перекрытия константы связи $J$ [Bredas2002].

Матрица будет полностью симметричной. Вы сможете вручную выписать:

* Элемент (1, 2) — полная внутристопочная связь $J_{\parallel}$.
* Элемент (1, 3) — полная межстопочная связь $J_{\perp}$.
* Элемент (1, 4) — полная диагональная связь межстопочного взаимодействия.

Проведение этого расчета вручную всего один раз позволит вам увидеть, как именно Multiwfn сопоставляет орбитали, и даст полное понимание структуры лога перед тем, как вы решите автоматизировать этот шаг на Python для массового скрининга.
