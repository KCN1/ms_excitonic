```python
import pandas as pd
```

```python
import py3Dmol
```

```python
df = pd.read_csv('ZnPc-gable-td035.gjf', header=None, sep=r'\s+')
df
```

```python
df[[1, 2, 3]] = df[[2, 3, 1]]
```

```python
viewer = py3Dmol.view()

viewer.addModel(str(len(df)) + '\nx\n' + df.to_string(header=None, index=False), 'xyz')

viewer.setStyle({'stick': {'radius': 0.1}})
viewer.zoomTo()
viewer.show()
```

```python
with open('1.xyz', 'w') as fp:
    print(df.to_string(formatters={
        0: '  {:2s}'.format,
    }, float_format='{:13.6f}'.format, header=None, index=False), file=fp)
```
