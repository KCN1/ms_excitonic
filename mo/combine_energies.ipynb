```python
import pandas as pd
from pathlib import Path
```

```python
def read_energies(filename):
    filename = Path(filename)
    frontier = pd.read_csv(filename, header=None, names=['symm', int(filename.stem.split('_')[-1])])
    frontier['occ'] = [2, 2, 0, 0, 0, 0]
    frontier.set_index(['occ', 'symm'], inplace=True)
    # frontier.index.names=[None, None]
    return frontier.T
```

```python
energies = pd.concat([read_energies(csvfile) for csvfile in Path.cwd().iterdir() if csvfile.is_file() and csvfile.suffix == '.csv']).sort_index()
```

```python
energies.to_excel('frontier.xlsx')
energies
```
